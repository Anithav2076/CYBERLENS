```python
def simulate_mitigation(sample, mitigation, model, original_risk):
    """
    Simulate one mitigation and estimate its effect on risk.
    """

    # Create a copy so the original system is not changed
    simulated = sample.copy()

    # Get the feature to modify
    feature = mitigation["feature"]

    # Get the reduction factor
    reduction_factor = mitigation["reduction_factor"]

    # Virtually modify the feature
    simulated[feature] = (
        simulated[feature] * reduction_factor
    )

    # Predict attack probability
    probability = model.predict_proba(simulated)[0, 1]

    # Convert probability to risk score
    new_risk = probability * 100

    # Calculate estimated reduction
    reduction = original_risk - new_risk

    return {
        "mitigation": mitigation["mitigation"],
        "feature": feature,
        "original_risk": original_risk,
        "new_risk": new_risk,
        "estimated_reduction": reduction
    }
```

```python
def simulate_recommendation(sample, recommendation, model, original_risk):
    """
    Simulate a recommendation and estimate its model-based risk impact.
    """

    # Copy the original system
    simulated = sample.copy()

    # Get feature and reduction
    feature = recommendation["feature"]
    
    # For this prototype, every scenario uses a 20% reduction
    reduction_factor = 0.80

    # Apply virtual feature change
    simulated[feature] = (
        simulated[feature] * reduction_factor
    )

    # Convert to NumPy because the model was trained on NumPy arrays
    simulated_array = simulated.to_numpy()

    # Predict attack probability
    probability = model.predict_proba(simulated_array)[0, 1]

    # Convert probability to risk score
    new_risk = probability * 100

    # Calculate estimated reduction
    reduction = original_risk - new_risk

    return {
        "mitigation_id": recommendation["mitigation_id"],
        "recommendation": recommendation["recommendation"],
        "feature": feature,
        "original_risk": original_risk,
        "new_risk": new_risk,
        "estimated_reduction": reduction,
        "security_rationale": recommendation["security_rationale"],
        "model_based_estimate": recommendation["model_based_estimate"]
    }
```

```python
def calculate_risk(probability):
    """
    Convert attack probability to a 0-100 risk score
    and assign a prototype risk level.
    """

    risk_score = probability * 100

    if risk_score < 25:
        risk_level = "LOW"
    elif risk_score < 50:
        risk_level = "MEDIUM"
    elif risk_score < 75:
        risk_level = "HIGH"
    else:
        risk_level = "CRITICAL"

    return risk_score, risk_level
```
